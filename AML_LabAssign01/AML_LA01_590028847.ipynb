## B1 — Scaling Changes the Answer

**Dataset:** California Housing

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)

knn_raw = KNeighborsRegressor(n_neighbors=5)
knn_raw.fit(X_train, y_train)

y_pred_raw = knn_raw.predict(X_test)
mae_raw = mean_absolute_error(y_test, y_pred_raw)

print(f"MAE without scaling: {mae_raw:.4f} ($100,000 units)")
print(f"Approximately ${mae_raw * 100000:,.0f}")
```

```python
knn_scaled = Pipeline([
    ("scaler", StandardScaler()),
    ("knn", KNeighborsRegressor(n_neighbors=5))
])

knn_scaled.fit(X_train, y_train)

y_pred_scaled = knn_scaled.predict(X_test)
mae_scaled = mean_absolute_error(y_test, y_pred_scaled)

print(f"MAE with scaling: {mae_scaled:.4f} ($100,000 units)")
print(f"Approximately ${mae_scaled * 100000:,.0f}")
```

### Observation (B1)

The MAE decreased from 0.8142 to 0.4308 after scaling. KNN uses distances,
so features with larger numerical scales can have greater influence on the
distance calculation. StandardScaler puts the features on a comparable scale,
which changes the nearest neighbours and improves the predictions.
